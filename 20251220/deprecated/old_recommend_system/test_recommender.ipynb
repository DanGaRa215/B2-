```python
from recommender import StoreRecommender
import pandas as pd

def main():
    # レコメンダーの初期化
    recommender = StoreRecommender()
    
    # テスト用クエリ
    query = "カジュアルでコスパが良い"
    print(f"\nQuery: {query}")
    print("="*80)

    # 4つのパターンで実験
    # A = 1.0 (純粋なイメージ重視)
    # A = 0.7 (イメージ重視 + 品質)
    # A = 0.3 (品質重視 + イメージ)
    # A = 0.0 (純粋な品質重視)
    alphas = [1.0, 0.7, 0.3, 0.0]
    
    for alpha in alphas:
        print(f"\n【Pattern A = {alpha}】 (Similarity: {int(alpha*100)}%, Rating: {int((1-alpha)*100)}%)")
        results = recommender.search(query, alpha=alpha, top_k=5)
        
        # 結果表示
        for i, result in enumerate(results, 1):
            print(f"\n{i}. {result['store_name']}")
            print(f"   URL: {result['store_url']}")
            print(f"   Score: {result['score']:.3f} | Similarity: {result['similarity']:.3f} | Rating: {result['rating']:.2f}")

if __name__ == "__main__":
    main()
```
